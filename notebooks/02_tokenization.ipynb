# Weird AI Tokenization Exploration

In this notebook we will:

- Load lyric data
- Explore the dataset
- Tokenize text
- Convert text into numbers
- Decode tokens back into text

This notebook supports Lesson 2 of the Weird AI project.

```python
from weird_ai.tokenizer import SimpleCharacterTokenizer
from pathlib import Path
from collections import Counter

text = Path("../data/processed/lyrics_sample.txt").read_text(
    encoding="utf-8"
)

def print_avg_length():
    lines = text.splitlines()
    non_empty_lines = [
          line.strip()
        for line in lines
        if line.strip()
    ]
    averagelength = (sum(len(line) for line in non_empty_lines) / len(non_empty_lines))
    print(f"Avg length: {averagelength:.2f}")

def count_unique_chars():
    tokenizer = SimpleCharacterTokenizer(text)
    print(f"{len(tokenizer.chars)} Unique chars")

def count_unique_words():
    words = set(text.split())
    print(f"{len(words)} Unique words")

def most_common_words():
    words = text.split()
    data = Counter(words).most_common(20)
    print(f"20 most common words: \n {data} ")



def print_all():
    print_avg_length()
    count_unique_chars()
    count_unique_words()
    most_common_words()

print_all()
```
