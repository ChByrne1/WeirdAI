# Lesson 16 – Mini-GRPO Training for Weird AI

This optional advanced lesson implements the missing reinforcement learning step in the Weird AI project.

## 1. Why This Lesson Exists

Earlier lessons created rewards and advantages, but they did not update the model using those rewards. This lesson adds the policy-gradient update.

## 2. The Mini-GRPO Pipeline

```text
Prompt → Generate rollouts → Evaluate → Rewards → Advantages → Log probabilities → Loss → Update weights
```

## 3. Rewards and Advantages

A reward says how good one rollout was. An advantage says how good it was compared with other rollouts from the same prompt.

````python
rewards = [0.0, 0.5, 1.0]
average = sum(rewards) / len(rewards)
advantages = [reward - average for reward in rewards]
print("rewards:", rewards)
print("advantages:", advantages)
````

## 4. Why Log Probabilities Matter

Generated text is not differentiable. The model learns from how probable it made that generated text.

````python
import torch
logprobs = torch.tensor([-1.0, -2.0, -3.0])
advantages = torch.tensor([1.0, 0.0, -1.0])
loss = -(advantages * logprobs).mean()
print(loss)
````

## 5. Policy-Gradient Loss

```python
loss = -(advantages * logprobs).mean()
```

Positive-advantage outputs should become more likely. Negative-advantage outputs should become less likely.

## 6. A Tiny Differentiable Example

````python
model = torch.nn.Linear(1, 1, bias=False)
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
features = torch.tensor([[0.3], [0.6], [0.9]])
logprobs = model(features).squeeze()
advantages = torch.tensor([-0.5, 0.0, 0.5])
loss = -(advantages * logprobs).mean()
optimizer.zero_grad()
loss.backward()
optimizer.step()
print("loss:", loss.item())
print("weight:", model.weight.data)
````

## 7. What This Mini Version Omits

Production GRPO usually adds KL penalties, clipped ratios, reference models, larger rollout batches, checkpoints, and extensive monitoring.
